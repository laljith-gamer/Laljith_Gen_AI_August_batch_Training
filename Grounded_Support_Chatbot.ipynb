```python
!pip install -q -U google-genai
from google import genai
from google.colab import userdata
```

```python
API_KEY = userdata.get("GEMINI_API_KEY")

if not API_KEY:
    raise ValueError(
        "GEMINI_API_KEY was not found. "
        "Add it to Google Colab Secrets first."
    )

client = genai.Client(api_key=API_KEY)

print("API key loaded successfully.")
```

```python
knowledge_base = """
Q: What are the CSBS department office timings?
A: The CSBS department office is open from 9:00 AM to 4:30 PM, Monday to Friday.

Q: What is the minimum attendance required for semester examination eligibility?
A: Students must maintain at least 75% attendance to be eligible for the semester examination.

Q: How many internal assessments are conducted in a semester?
A: Three internal assessments are conducted during each semester.

Q: How many copies of the final project report are required?
A: Two printed copies of the final project report are required.

Q: When should the final project report be submitted?
A: The final project report should be submitted 7 days before the project viva.

Q: How long does it normally take to receive a bonafide certificate?
A: A bonafide certificate is normally issued within 2 working days.

Q: How long can students use laboratory equipment?
A: Laboratory equipment can be borrowed for a maximum of 2 hours during a laboratory session.

Q: Can students use the computer lab after regular class hours?
A: Students can use the computer lab after class hours only with department permission.

Q: Where should internship documents be submitted?
A: Internship documents should be submitted to the CSBS department office.

Q: What should I do if I have a problem with my internal marks?
A: Students should first contact the concerned subject faculty. If the issue is not resolved, they should contact the department office.

Q: What should I do if I lose my college ID card?
A: Students should report the lost ID card to the department office and apply for a replacement.

Q: Can attendance corrections be requested after the semester ends?
A: Attendance corrections must be reported to the department office before the attendance closing date.

Q: How early should I request a department certificate?
A: Certificate requests should normally be submitted at least 2 working days before the certificate is required.

Q: When should students meet the HOD?
A: Students should normally request an appointment with the department office before meeting the HOD.

Q: What are the department working days?
A: The department works from Monday to Friday.
"""

print("Knowledge base loaded.")
print("Number of characters:", len(knowledge_base))
```

```python
fallback_message = (
    "I don't have that information in my knowledge base. "
    "Please contact the CSBS department office at "
    "044-12345678 or csbs.office@example.com."
)

system_instruction = f"""
You are the CSBS Department Office Support Bot.

Your job is to answer questions about the CSBS department office
using ONLY the knowledge base provided below.

RULES:

1. Answer only from the knowledge base.
2. Do not use outside knowledge.
3. Do not guess.
4. Do not invent facts, prices, dates, rules, timings, limits, or procedures.
5. If the answer is not available in the knowledge base, use exactly this sentence:

"{fallback_message}"

6. Keep answers clear and concise.
7. You may use information from previous messages in the conversation.
8. Do not contradict the knowledge base.
9. If the user asks a question containing multiple parts, answer only the parts
   that are supported by the knowledge base.
10. Never pretend that information exists in the knowledge base when it does not.

KNOWLEDGE BASE:

{knowledge_base}
"""

print(system_instruction)
```

```python
chat = client.chats.create(
    model="gemini-3.5-flash",
    config={
        "temperature": 0.2,
        "system_instruction": system_instruction
    }
)

print("CSBS Support Bot is ready!")
```

```python
while True:

    user_input = input("You: ").strip()

    if user_input.lower() in ["quit", "exit"]:
        print("Bot: Goodbye!")
        break

    if not user_input:
        print("Bot: Please enter a question.")
        continue

    try:

        response = chat.send_message(user_input)

        print("Bot:", response.text)
        print()

    except Exception as e:

        print("Bot: An error occurred")
        print("Error:", e)
```
