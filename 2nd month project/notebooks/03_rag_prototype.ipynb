# SmartHire GenAI — 03: Career Mentor RAG Prototype
Prototypes grounded question answering over the career notes corpus with strict anti-hallucination refusal.

```python
import sys
sys.path.append('..')
from src.mentor.rag_chain import CareerMentorRAG
from src.safety.guardrails import SafetyGuardrails

mentor = CareerMentorRAG()
print('Career Mentor RAG initialized.')
```

```python
query = 'How do I transition from software engineering into Machine Learning?'
safe, reason = SafetyGuardrails.evaluate_input(query)
print(f'Guardrail check: {"SAFE" if safe else f"BLOCKED: {reason}"}')

if safe:
    response = mentor.answer_question(query)
    print('\n--- Grounded Answer ---\n', response.answer)
    print('\n--- Citations ---\n', [c.document_title for c in response.citations])
```

```python
unsupported_query = 'What is the exact internal travel reimbursement limit for Acme Widgets in 2029?'
resp = mentor.answer_question(unsupported_query)
print('Unsupported Query Answer:\n', resp.answer)
refused = any(term in resp.answer.lower() for term in ["don't know", 'not found', 'does not contain'])
print('Strict refusal enforced:', refused)
```
