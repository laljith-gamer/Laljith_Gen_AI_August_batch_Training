# SmartHire GenAI — 02: Build FAISS Vector Index
Loads the job postings corpus and builds/queries the FAISS vector index.

```python
import sys
sys.path.append('..')
import pandas as pd
from src.config import settings
from src.search.job_search import JobSearchEngine

jobs_df = pd.read_csv(settings.JOBS_DATA_PATH)
print(f'Loaded {len(jobs_df)} jobs from {settings.JOBS_DATA_PATH}')
jobs_df.head(3)
```

```python
engine = JobSearchEngine()
print(f'FAISS index loaded. Total vectors: {engine.index.ntotal if engine.index else 0}')
```

```python
query = 'Python Backend Engineer with FastAPI, PostgreSQL, and Docker'
results = engine.search_by_text(query, top_k=3)

for idx, m in enumerate(results):
    print(f'[{idx+1}] {m.job.title} at {m.job.company} (Score: {m.similarity_score:.3f})')
```
