# SmartHire GenAI — 01: Embeddings Exploration
This notebook tests text embeddings generation using Google Gemini and evaluates cosine similarity between queries and job titles.

```python
import sys
sys.path.append('..')
import numpy as np
from src.search.embed import EmbeddingEngine

engine = EmbeddingEngine()
print('Embedding engine initialized.')
```

```python
texts = [
    'Senior Machine Learning Engineer specializing in PyTorch and LLM fine-tuning',
    'AI/ML Research Scientist working on Generative AI and diffusion models',
    'Lead Frontend Developer experienced in React, TypeScript, and CSS'
]

vectors = engine.embed_texts(texts)
print(f'Generated {len(vectors)} embeddings of dimension: {len(vectors[0])}')
```

```python
def cosine_sim(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

sim_ml_ai = cosine_sim(vectors[0], vectors[1])
sim_ml_frontend = cosine_sim(vectors[0], vectors[2])

print(f'Similarity (ML Eng vs AI Scientist): {sim_ml_ai:.4f}')
print(f'Similarity (ML Eng vs Frontend):     {sim_ml_frontend:.4f}')
assert sim_ml_ai > sim_ml_frontend, 'ML roles must have higher similarity than ML vs Frontend'
```
